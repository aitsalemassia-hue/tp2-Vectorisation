```python

from sklearn.feature_extraction.text import CountVectorizer  
corpus = ["J'aime programmer en Python pour analyser les données",
    "Le langage Python est très utilisé pour l'intelligence artificielle et les données",
    "Les réseaux informatiques permettent de partager les données entre les ordinateurs"
]
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(corpus)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(corpus)
print(vectorizer.get_feature_names_out())
print(X.toarray())
```
